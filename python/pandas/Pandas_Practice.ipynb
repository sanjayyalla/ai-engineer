```python
import pandas as pd

employees = pd.DataFrame({
    "EmployeeID":[1,2,3,4],
    "Name":["Sanjay","Rahul","Priya", "Amit"],
    "DepartmentID":[101,102,101,99]
})


departments = pd.DataFrame({
    "DepartmentID":[101,102],
    "Department":["AI","Cloud"]
})
```

```python
concatenated_df = pd.concat([employees, departments],axis=0)
concatenated_df
```

```python
merged_df = pd.merge(employees, departments, on="DepartmentID",how="inner")
merged_df
```

```python
pd.merge(employees, departments, on="DepartmentID", how="left")
```

```python
pd.merge(employees, departments, on="DepartmentID", how="right")
```

```python
pd.merge(employees, departments, on="DepartmentID", how="outer")
```

```python
import pandas as pd
users = [ { "Name": "Sanjay", "Age": 24, "City": "Hyderabad" }, { "Name": "Rahul", "Age": 22, "City": "Delhi" } ]
df = pd.DataFrame(users)

df.to_json("output.json", orient="records") #We have multiple orient options available for orient parameter. The default is 'columns', which will create a JSON object with column names as keys and lists of column values as values. The 'records' option will create a list of dictionaries, where each dictionary represents a row in the DataFrame.
```

```python
data = { "Department": "AI", "Employees": [ {"Name": "Sanjay"}, {"Name": "Rahul"} ] }
```

```python
pd.json_normalize(data,record_path="Employees",meta=["Department"])
```

```python
import numpy as np
import time

python_list = list(range(1_000_000))
numpy_array = np.arange(1_000_000)

start = time.time()
python_result = [x * 2 for x in python_list]
print(f"Python list: {time.time() - start:.4f}s")

start = time.time()
numpy_result = numpy_array * 2
print(f"NumPy array: {time.time() - start:.4f}s")
```

```python
import time

df = pd.DataFrame({"text": ["hello world"] * 100_000})

start = time.time()
df["text"].apply(len)              # disguised loop
print(f"apply: {time.time() - start:.4f}s")

start = time.time()
df["text"].str.len()                # vectorized
print(f"str.len: {time.time() - start:.4f}s")
```
