```python
import numpy as np

documents = np.array([
    [3, 4],
    [4, 3],
    [1, 0]
])

query = np.array([3.5, 3.5])

# Normalize documents
document_norms = np.linalg.norm(
    documents,
    axis=1,
    keepdims=True
)
```

```python
document_norms
```

```python
documents_normalized = documents / document_norms
documents_normalized
```

```python
query_normalized = query / np.linalg.norm(query)
query_normalized
```

```python
scores = documents_normalized @ query_normalized
scores
```

```python
rankings = np.argsort(scores)[::-1]
rankings
```

```python
def softmax(x: np.ndarray) -> np.ndarray:
    exp_x = np.exp(x - np.max(x))   # subtract max for numerical stability
    print("exp_x:", exp_x)
    return exp_x / np.sum(exp_x)

logits = np.array([11.0, 1999943.0, 0.1])
res= softmax(logits)   
```
